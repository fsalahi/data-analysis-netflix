# Phase 2: Cleaning Data

```python
# Make a copy of the original data
df_clean = df.copy()
```

```python
df_clean.isnull().sum()
```

```python
# Let'd decide about some of these missing values
# this dataset is mostly text so replacing missing values with a default text makes sense
# if the value was numerical we could .fillna(mean) or .fillna(median) based on the logic of situation
df_clean["director"] = df_clean["director"].fillna("Unknown")
```

```python
df_clean.dtypes
```

```python
df_clean["date_added"] = pd.to_datetime(df_clean["date_added"], format="mixed")
```

In the above cell at first I didn't pass the format argument, it threw an error about the format of a few records that it couldn't match. with this argument I stated that there could be inconsistancies in different rows of this value and expected it to comprehend this. 
In real project I should keep record of these records and make sure the data/time is converted accordingly. As another approach I may use another format to enforce a specific conversion.

```python
df_clean.dtypes
```

## Feature engineering
Let's add year, month, and day of adding the data from the date_added

```python
df_clean["year_added"] = df_clean["date_added"].dt.year
df_clean["month_added"] = df_clean["date_added"].dt.month_name()
df_clean["day_added"] = df_clean["date_added"].dt.day_name()
```

```python
df_clean.sample(10)
```

Now let's address the inconsistancies in the country.

```python
df_clean["country"].value_counts().head(50)
# so we see in addition to the reasons we named above causing to have too many distinct countries
# another reason is there is movies made by more than one country
```

```python
#first let's clear hidden spaces
df_clean["country"] = df_clean["country"].str.strip()
```

```python
df_clean["country"].nunique()
```

distinct values didn't change: 748, so space wasn't the problem

```python
# let's deal with case sensitivity
df_clean["country"] = df_clean["country"].str.lower()
```

```python
df_clean["country"].nunique()
```

```python
#let's do these for all columns
#find all text-based columns
string_cols = df_clean.select_dtypes(include=['object', 'string']).columns
df_clean[string_cols] = df_clean[string_cols].apply(lambda x: x.str.lower())
df_clean[string_cols] = df_clean[string_cols].apply(lambda x: x.str.strip())
```

```python
df_clean.info()
```

```python
df_clean.columns
```

```python
#to increse readability of "listed_in". it makes more sense if it is named "genres"
df_clean.rename(
    columns={
        "listed_in": "genres"
    },
    inplace=True
)
```
